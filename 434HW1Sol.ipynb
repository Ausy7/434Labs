## Guassian Distribution

I chose a mean of 3 and a std dev of 0.5. Mean sets center, std dev sets width.

The probability density is

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
\exp\left[-\frac{(x-\mu)^2}{2\sigma^2}\right].
$$

with my values,

$$
f(x)=\frac{1}{0.5\sqrt{2\pi}}
\exp\left[-\frac{(x-3)^2}{0.5}\right].
$$

as above, I generated 100k random samples and plotted them in a histogram of 50 bins. The histogram is normalized so total area is 1.

```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams["figure.figsize"] = (10, 6)

mu = 3.0
sigma = 0.5

#generate reproducible random samples.
rng = np.random.default_rng(42)
d = stats.norm.rvs(
    loc=mu, scale=sigma, size=100000, random_state=rng
)

#calculate the theoretical density.
limits = (mu - 5 * sigma, mu + 5 * sigma)
x = np.linspace(*limits, 1000)
pdf = stats.norm.pdf(x, loc=mu, scale=sigma)

#use the same bins for all three plots.
bins = np.linspace(*limits, 51)

fig, ax = plt.subplots()
ax.hist(d, bins=bins, density=True, alpha=0.6,
        label="Random samples")
ax.plot(x, pdf, linewidth=3, label="Normal density")
ax.set(xlim=limits, xlabel="x", ylabel="Probability density",
       title="Gaussian distribution: mean 3, standard deviation 0.5")
ax.legend()
plt.show()
```

The log scale makes low density tails easier to see, so Guassian curve looks like downward facing parabola because

$$
\ln f(x)=-\ln(\sigma\sqrt{2\pi})
-\frac{(x-\mu)^2}{2\sigma^2}.
$$

Histogram fluctuates more in the tails because fewer samples fall there. empty bins cant be shown because log(0) is undefined!

```python
fig, ax = plt.subplots()
ax.hist(d, bins=bins, density=True)
ax.set_yscale("log")
ax.set(xlim=limits, xlabel="x", ylabel="Probability density (log scale)",
       title="Gaussian histogram on a logarithmic scale")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.hist(d, bins=bins, density=True, alpha=0.6,
        label="Random samples")
ax.plot(x, pdf, linewidth=3, label="Normal density")
ax.set_yscale("log")
ax.set(xlim=limits, xlabel="x", ylabel="Probability density (log scale)",
       title="Logarithmic histogram with normal density")
ax.legend()
plt.show()
```
